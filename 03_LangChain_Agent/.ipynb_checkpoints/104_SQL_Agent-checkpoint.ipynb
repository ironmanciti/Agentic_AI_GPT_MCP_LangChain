# SQL Agent (SQL 데이터베이스 Q&A 에이전트)

이 튜토리얼에서는 LangChain을 사용하여 SQL 데이터베이스에 대한 자연어 질의응답 시스템을 구축합니다.
에이전트는 자연어 질문을 SQL 쿼리로 변환하고, 결과를 바탕으로 답변을 생성합니다.

**참고**: [LangChain 공식 문서 - SQL Agent](https://docs.langchain.com/oss/python/langchain/sql-agent)

## 학습 내용
- SQL 데이터베이스 연결 (읽기 전용)
- `@tool`로 SQL 도구 직접 정의 (실행 전 쿼리 검증)
- SQL 에이전트 구축
- Human-in-the-loop 검토

````python
from dotenv import load_dotenv
import os

load_dotenv()
````

````python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    os.environ["OPENAI_MODEL"],
    model_provider="openai",
    use_responses_api=True,        # Responses API 사용 (Chat Completions에서는 추론 모델의 도구 호출 불가)
    reasoning={"effort": "low"},   # 추론 강도 낮춤 → 실습 속도·비용 절감
)
````

## 1. 데이터베이스 연결

파이썬 내장 `sqlite3`로 SQLite 데이터베이스에 **읽기 전용**으로 연결합니다. 이 예제에서는 Chinook 샘플 데이터베이스를 사용합니다.

### 1.1 Chinook 데이터베이스 준비

Chinook은 디지털 미디어 스토어를 나타내는 샘플 데이터베이스입니다.

데이터베이스가 없으면 다음 명령어로 생성할 수 있습니다:
```bash
curl -s https://raw.githubusercontent.com/lerocha/chinook-database/master/ChinookDatabase/DataSources/Chinook_Sqlite.sql | sqlite3 Chinook.db
```

````python
import sqlite3

DB_PATH = "Chinook.db"

def connect_readonly() -> sqlite3.Connection:
    """DB를 읽기 전용(mode=ro)으로 연결 — 에이전트가 만든 SQL이 데이터를 바꾸지 못하도록 권한을 최소로 제한
    (파일이 없으면 빈 DB를 새로 만들지 않고 바로 에러가 남)"""
    return sqlite3.connect(f"file:{DB_PATH}?mode=ro", uri=True)


# 데이터베이스 정보 확인
con = connect_readonly()
try:
    tables = [row[0] for row in con.execute(
        "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';"
    )]
    print("사용 가능한 테이블:")
    print(tables)

    # 샘플 쿼리 실행
    sample_result = con.execute("SELECT * FROM Artist LIMIT 5;").fetchall()
    print(f"\n샘플 쿼리 결과:\n{sample_result}")
finally:
    con.close()
````

## 2. SQL 도구 생성

최신 공식 튜토리얼처럼 `@tool` 데코레이터로 SQL 도구 4개를 직접 정의합니다.

| 도구 | 역할 |
|------|------|
| `sql_db_list_tables` | 테이블 목록 조회 |
| `sql_db_schema` | 테이블 스키마(CREATE 문) + 샘플 3행 조회 |
| `sql_db_query` | SQL 실행 (SELECT 조회만 허용) |
| `sql_db_query_checker` | 실행 전에 LLM으로 쿼리의 흔한 실수 점검 |

**보안**: 모델이 만든 SQL을 그대로 실행하는 것은 위험합니다. 공식 문서 권고에 따라 두 겹으로 방어합니다.
1. **읽기 전용 연결** (`mode=ro`) — 권한을 최소로 제한
2. **실행 전 검증** (`validate_select`) — SELECT/WITH로 시작하는 단일 문장만 허용

````python
from langchain.tools import tool

def validate_select(query: str) -> str:
    """조회(SELECT/WITH) 단일 문장만 허용하고, 아니면 ValueError 발생
    (WITH 로 시작하는 쓰기 문장은 읽기 전용 연결이 2차로 막음)"""
    q = query.strip().rstrip(";").strip()
    if ";" in q:
        raise ValueError("여러 SQL 문장은 한 번에 실행할 수 없습니다.")
    if not q.lower().startswith(("select", "with")):
        raise ValueError("SELECT(또는 WITH) 조회 쿼리만 실행할 수 있습니다.")
    return q

@tool
def sql_db_list_tables() -> str:
    """데이터베이스의 테이블 목록을 쉼표로 구분해 반환합니다."""
    con = connect_readonly()
    try:
        rows = con.execute(
            "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';"
        ).fetchall()
        return ", ".join(row[0] for row in rows)
    finally:
        con.close()

@tool
def sql_db_schema(table_names: str) -> str:
    """쉼표로 구분한 테이블 이름들을 받아 각 테이블의 스키마(CREATE 문)와 샘플 3행을 반환합니다.
    먼저 sql_db_list_tables 로 테이블이 실제로 있는지 확인하세요.
    예) Artist, Album"""
    con = connect_readonly()
    try:
        valid_tables = {row[0] for row in con.execute("SELECT name FROM sqlite_master WHERE type='table';")}
        results = []
        for table in (name.strip() for name in table_names.split(",")):
            if table not in valid_tables:
                results.append(f"Error: '{table}' 테이블이 데이터베이스에 없습니다.")
                continue
            (create_sql,) = con.execute(
                "SELECT sql FROM sqlite_master WHERE type='table' AND name=?;", (table,)
            ).fetchone()
            # table 은 위에서 실제 테이블 이름인지 확인했으므로 쿼리에 넣어도 안전
            cursor = con.execute(f'SELECT * FROM "{table}" LIMIT 3;')
            columns = "\t".join(col[0] for col in cursor.description)
            rows = "\n".join("\t".join(str(v) for v in row) for row in cursor.fetchall())
            results.append(f"{create_sql}\n/*\n{table} 테이블 샘플 3행:\n{columns}\n{rows}\n*/")
        return "\n\n".join(results)
    finally:
        con.close()

@tool
def sql_db_query(query: str) -> str:
    """SQL 조회 쿼리를 실행하고 결과를 반환합니다. (SELECT 문만 허용)
    쿼리가 잘못되면 에러 메시지가 반환되니, 쿼리를 고쳐서 다시 시도하세요.
    'no such column' 에러가 나면 sql_db_schema 로 컬럼 이름을 확인하세요."""
    try:
        q = validate_select(query)          # 1차 방어: 조회 쿼리만 허용
        con = connect_readonly()            # 2차 방어: 읽기 전용 연결
        try:
            return str(con.execute(q).fetchall())
        finally:
            con.close()
    except Exception as e:
        return f"Error: {e}"

@tool
def sql_db_query_checker(query: str) -> str:
    """쿼리를 실행하기 전에 흔한 실수가 있는지 점검합니다.
    sql_db_query 로 실행하기 전에 항상 이 도구를 먼저 사용하세요."""
    prompt = f"""{query}
위 SQLite 쿼리에 다음과 같은 흔한 실수가 있는지 확인하세요:
- NULL 값이 있는 컬럼에 NOT IN 사용
- UNION ALL 을 써야 할 곳에 UNION 사용
- 배타적 범위에 BETWEEN 사용
- 조건절의 데이터 타입 불일치
- 식별자 따옴표 처리
- 함수 인자 개수
- 올바른 타입으로의 형 변환
- 조인에 올바른 컬럼 사용

실수가 있으면 쿼리를 다시 작성하고, 없으면 원래 쿼리를 그대로 출력하세요.
최종 SQL 쿼리만 출력하세요.

SQL Query: """
    return model.invoke(prompt).text.strip()

tools = [sql_db_list_tables, sql_db_schema, sql_db_query, sql_db_query_checker]

# 사용 가능한 도구 확인 (루프 변수를 t 로 써서 @tool 데코레이터 이름을 가리지 않음)
print("사용 가능한 SQL 도구:")
for t in tools:
    print(f"\n- {t.name}")
    print(f"  설명: {t.description}")
````

## 3. SQL 에이전트 생성

SQL 도구를 사용하는 에이전트를 생성합니다.

````python
from langchain.agents import create_agent

# 시스템 프롬프트 설정
system_prompt = """
당신은 SQL 데이터베이스와 상호작용하도록 설계된 에이전트입니다.
사용자로부터 질문을 입력받으면, 실행 가능한 {dialect} 구문으로
문법적으로 올바른 SQL 쿼리를 생성하세요.

그 후 쿼리를 실행한 결과를 확인하고, 그에 기반하여 답변을 반환하세요.
사용자가 명시적으로 원하는 예시 개수를 지정하지 않은 경우,
항상 최대 {top_k}개의 결과만 반환하도록 쿼리를 제한해야 합니다.

데이터베이스에서 가장 흥미로운 예시를 보여주기 위해,
관련 있는 열(column)을 기준으로 결과를 정렬할 수 있습니다.
특정 테이블의 모든 열을 조회하지 말고, 질문에 필요한 관련 열만 선택하세요.

쿼리를 실행하기 전에 반드시 구문을 두 번 확인하세요.
만약 쿼리 실행 중 오류가 발생하면, 쿼리를 다시 작성하고 재시도해야 합니다.

데이터베이스에 DML 문(INSERT, UPDATE, DELETE, DROP 등)을
절대 실행하지 마세요.

항상 데이터베이스에 어떤 테이블들이 존재하는지 먼저 확인한 뒤
어떤 데이터를 조회할 수 있는지 파악해야 합니다.
이 단계를 건너뛰지 마세요.

그 다음, 가장 관련성 높은 테이블의 스키마(schema)를 조회하세요.
""".format(
    dialect="SQLite",
    top_k=5,
)

# SQL 에이전트 생성
agent = create_agent(
    model,
    tools,
    system_prompt=system_prompt,
)
````

## 4. 에이전트 실행

자연어 질문으로 데이터베이스를 쿼리합니다.

````python
# 질문 1: 집계 쿼리
question1 = "어떤 장르가 평균적으로 가장 긴 트랙을 가지고 있나요?"

print(f"질문: {question1}\n")
print("=" * 80)

for step in agent.stream(
    {"messages": [{"role": "user", "content": question1}]},
    stream_mode="values",
):
    step["messages"][-1].pretty_print()
````

````python
# 질문 2: 테이블 설명
question2 = "playlisttrack 테이블의 구조를 설명해주세요."

print(f"\n질문: {question2}\n")
print("=" * 80)

for step in agent.stream(
    {"messages": [{"role": "user", "content": question2}]},
    stream_mode="values",
):
    step["messages"][-1].pretty_print()
````

````python
# 질문 3: 복잡한 조인 쿼리
question3 = "가장 많은 트랙을 가진 아티스트는 누구인가요?"

print(f"\n질문: {question3}\n")
print("=" * 80)

for step in agent.stream(
    {"messages": [{"role": "user", "content": question3}]},
    stream_mode="values",
):
    step["messages"][-1].pretty_print()
````
